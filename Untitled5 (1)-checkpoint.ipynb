```python
import cv2
import numpy as np
import os
import pandas as pd

# Set the BRIEF and KAZE parameters
brief_params = dict(bytes=32, use_orientation=False)
kaze_params = dict(threshold=0.001, nOctaves=4)

# Initialize the BRIEF and KAZE extractors
brief_extractor = cv2.xfeatures2d.BriefDescriptorExtractor_create(**brief_params)
kaze_detector = cv2.KAZE_create(**kaze_params)

# Loop over the images in the dataset
image_dir = r"C:\Users\Madhuri-Amit\Desktop\preprocessed"
descriptors_list = []
labels_list = []
for class_name in os.listdir(image_dir):
    class_dir = os.path.join(image_dir, class_name)
    for image_filename in os.listdir(class_dir):
        # Load the image and convert it to grayscale
        image_path = os.path.join(class_dir, image_filename)
        image = cv2.imread(image_path)
        gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

        # Detect keypoints with the KAZE detector
        keypoints = kaze_detector.detect(gray_image)

        # Compute BRIEF descriptors for the keypoints
        keypoints, descriptors = brief_extractor.compute(gray_image, keypoints)

        # Concatenate the BRIEF and KAZE descriptors
        if descriptors is not None:
            kaze_descriptors = np.array([keypoint.response for keypoint in keypoints])
            combined_descriptors = np.concatenate((descriptors, kaze_descriptors.reshape(-1, 1)), axis=1)

            # Append the combined descriptors to the list
            descriptors_list.append(np.mean(combined_descriptors, axis=0))

            # Append the class label to the list
            labels_list.append(class_name)

# Convert the descriptor list and label list to Pandas dataframes
descriptors_df = pd.DataFrame(np.vstack(descriptors_list))
labels_df = pd.DataFrame({"label": labels_list})

# Concatenate the descriptor and label dataframes and save to a CSV file
combined_df = pd.concat([descriptors_df, labels_df], axis=1)
combined_df.to_csv("final3.csv", index=False)

# Print the number of features obtained
num_features = descriptors_df.shape[1]
print("Number of columns: ", num_features)
```
